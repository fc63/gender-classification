```python
from datasets import load_dataset
import pandas as pd
import matplotlib.pyplot as plt

# Luna datasetini yükle
luna_ds = load_dataset("czyzi0/luna-speech-dataset", split="train")
# DataFrame'e çevir ve sadece gerekli sütunları al
luna_df = luna_ds.to_pandas()
luna_df = luna_df[['gender', 'transcript']]
luna_df = luna_df[luna_df['gender'].isin(['male', 'female'])]

# PWR AZON datasetini yükle
pwr_ds = load_dataset("czyzi0/pwr-azon-speech-dataset", split="train")
# DataFrame'e çevir ve sadece gerekli sütunları al
pwr_df = pwr_ds.to_pandas()
pwr_df = pwr_df[['gender', 'transcript']]
pwr_df = pwr_df[pwr_df['gender'].isin(['male', 'female'])]

# İki dataframe'i birleştir
df = pd.concat([luna_df, pwr_df], ignore_index=True)

# Görselleştir (df yazınca tablo olarak çıksın)
import seaborn as sns
plt.figure(figsize=(8, 4))
sns.countplot(data=df, x='gender')
plt.title("Gender Dağılımı")
plt.show()

# df içeriğini yazdır
df
```
