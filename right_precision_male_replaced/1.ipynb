```python
# kütüphaneler
import pandas as pd
import numpy as np
import random
from google.colab import drive

# google drive'ı mount et
drive.mount('/content/drive')

# tekrar eden satırları sil: correct_predicted_males.pkl
correct_males_path = "/content/drive/MyDrive/datasets/correct_predicted_males.pkl"
correct_males_df = pd.read_pickle(correct_males_path)
correct_males_df = correct_males_df.drop_duplicates()

# right_precision.pkl dosyasını yükle ve cinsiyet dengesini yazdır
right_precision_path = "/content/drive/MyDrive/datasets/right_precision.pkl"
right_df = pd.read_pickle(right_precision_path)
print("Orijinal right_precision.pkl cinsiyet dağılımı:")
print(right_df["gender"].value_counts())

# sadece 'male' satırları al
male_indices = right_df[right_df["gender"] == "male"].index.tolist()

# örnekleri rastgele değiştir (eşit sayıdaysa birebir, fazlaysa örnekle)
num_to_replace = min(len(male_indices), len(correct_males_df))
replacement_samples = correct_males_df.sample(n=num_to_replace, random_state=42).reset_index(drop=True)

# right_df'nin kopyasını oluştur
modified_df = right_df.copy()

# belirlenen indekslerdeki satırları değiştir
for i, idx in enumerate(random.sample(male_indices, num_to_replace)):
    modified_df.loc[idx] = replacement_samples.loc[i]

# yeni dengeyi yazdır
print("\nYeni dataset cinsiyet dağılımı:")
print(modified_df["gender"].value_counts())

# yeni dataset kaydedilsin
output_path = "/content/drive/MyDrive/datasets/right_precision_male_replaced.pkl"
modified_df.to_pickle(output_path)
print(f"\nYeni dataset kaydedildi: {output_path}")
```
