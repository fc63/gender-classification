```python
import pickle
import pandas as pd

# dosya yolları
path1 = "/content/drive/MyDrive/datasets/europarl_normalized.pkl"
path2 = "/content/drive/MyDrive/datasets/combinedatasetinformal.pkl"

# yükle
with open(path1, "rb") as f1:
    df1 = pickle.load(f1)

with open(path2, "rb") as f2:
    df2 = pickle.load(f2)

# birleştir
df_merged = pd.concat([df1, df2], ignore_index=True)

# kaydet
output_path = "/content/drive/MyDrive/datasets/merged_europarl_informal.pkl"
with open(output_path, "wb") as f_out:
    pickle.dump(df_merged, f_out)
```
