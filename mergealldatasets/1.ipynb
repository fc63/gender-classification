```python
import pickle
import pandas as pd

# dosya yolları
path1 = "/content/drive/MyDrive/datasets/europarl_normalized.pkl"
path2 = "/content/drive/MyDrive/datasets/combinedatasetinformal.pkl"

# yükle
with open(path1, "rb") as f1:
    df1 = pickle.load(f1)

with open(path2, "rb") as f2:
    df2 = pickle.load(f2)
```

```python
# birleştir
df_merged = pd.concat([df1, df2], ignore_index=True)
```

```python
# Orijinal veri kopyalanır
df_original = df_merged.copy()

# Dengeleme işlemi yapılır
min_count = df_original['gender'].value_counts().min()
df_balanced = (
    df_original.groupby('gender', group_keys=False)
    .apply(lambda x: x.sample(n=min_count, random_state=63))
    .reset_index(drop=True)
)

# Çıkarılan verileri bul
df_removed = pd.concat([df_original, df_balanced]).drop_duplicates(keep=False)
```

```python
# kaydet
output_path = "/content/drive/MyDrive/datasets/merged_europarl_informal.pkl"
with open(output_path, "wb") as f_out:
    pickle.dump(df_balanced, f_out)
```

```python
# kaydet
output_path = "/content/drive/MyDrive/datasets/removeedilenmale.pkl"
with open(output_path, "wb") as f_out2:
    pickle.dump(df_removed, f_out2)
```
