```python
import re
import pandas as pd

def remove_chat_logs(df, text_column):
    """
    Blog veri setinden chat log tarzı içerikleri temizler.
    text_column: Temizlenecek yazının bulunduğu sütunun adı
    """
    def is_chat_log(text):
        # 2 veya daha fazla "Saat (AM/PM)" desenini içeriyorsa sohbet olabilir
        time_pattern = r'\d{1,2}:\d{2}\s?(?:AM|PM)'
        matches = re.findall(time_pattern, text, flags=re.IGNORECASE)
        return len(matches) >= 2

    df_cleaned = df[~df[text_column].apply(is_chat_log)].copy()
    return df_cleaned

df = remove_chat_logs(df, text_column="text")

df
```

```python
def remove_media_caption_rows(df, text_column):
    """
    urlLink gibi medya placeholder'ı içeren, genelde bilgi taşımayan satırları kaldırır.
    """
    pattern = r'^urlLink\s.*\surlLink$'
    df_cleaned = df[~df[text_column].str.match(pattern, case=False, na=False)].copy()
    return df_cleaned

df = remove_media_caption_rows(df, text_column="text")

df
```

```python
def remove_chat_format_rows(df, text_column):
    """
    Chat log formatında yazılmış satırları kaldırır.
    - Birden fazla timestamp varsa (çoklu mesaj)
    - Veya başında "Username (HH:MM AM/PM):" formatı varsa (tekli mesaj)
    """
    # çoklu timestamp kontrolü (önceki sistem)
    multi_time_pattern = r'(\d{1,2}:\d{2}\s?(?:AM|PM))'

    # tekli chat formatı: Username (HH:MM AM/PM):
    single_chat_pattern = r'^[^\n]{1,40}\s\(\d{1,2}:\d{2}\s?(?:AM|PM)\):'

    def is_chat_line(text):
        time_matches = re.findall(multi_time_pattern, text, flags=re.IGNORECASE)
        single_line_match = re.match(single_chat_pattern, text, flags=re.IGNORECASE)
        return len(time_matches) >= 2 or bool(single_line_match)

    df_cleaned = df[~df[text_column].apply(is_chat_line)].copy()
    df_cleaned.reset_index(drop=True, inplace=True)
    return df_cleaned

df = remove_chat_format_rows(df, text_column="text")

df
```

```python
import re

def is_likely_quoted(text):
    if len(text) < 300:
        return False

    # tırnak içinde uzun metin
    if re.match(r'^\s*["“].+["”]\s*$', text, re.DOTALL):
        return True

    # yaygın quote keyword'leri
    keywords = ['by ', 'source:', 'quoted from', 'excerpt', 'taken from', '--', '~', 'via ', 'posted by']
    if any(kw in text.lower() for kw in keywords):
        return True

    return False

df = df[~df["text"].apply(is_likely_quoted)].copy().reset_index(drop=True)

df
```

```python
import re

def clean_noisy_rows(df, text_column="text", remove_short=True, min_length=15):
    def is_chat_log(text):
        # çoklu timestamp ya da Username (HH:MM AM/PM): şeklinde
        multi_time_pattern = r'(\d{1,2}:\d{2}\s?(?:AM|PM))'
        single_line_pattern = r'^[^\n]{1,40}\s\(\d{1,2}:\d{2}\s?(?:AM|PM)\):'
        return len(re.findall(multi_time_pattern, text, flags=re.IGNORECASE)) >= 2 or \
               re.match(single_line_pattern, text, flags=re.IGNORECASE)

    def is_urllink_line(text):
        return text.lower().count("urllink") >= 2 or text.lower().startswith("urllink")

    def is_quoted_line(text):
        if len(text) < 300:
            return False
        if re.match(r'^\s*["“].+["”]\s*$', text, re.DOTALL):
            return True
        keywords = ['by ', 'source:', 'quoted from', 'excerpt', 'taken from', '--', '~', 'via ', 'posted by']
        return any(kw in text.lower() for kw in keywords)

    def is_too_short(text):
        return len(text.strip()) < min_length

    def is_noisy(text):
        return is_chat_log(text) or is_urllink_line(text) or is_quoted_line(text) or (remove_short and is_too_short(text))

    df_cleaned = df[~df[text_column].apply(is_noisy)].copy()
    df_cleaned.reset_index(drop=True, inplace=True)
    return df_cleaned

df = clean_noisy_rows(df, text_column="text")

df
```

```python
df["text"] = df["text"].str.replace(r"\burlLink\b", "", regex=True)
df = df[~df["text"].str.startswith("mood:", na=False)].copy()

def is_useless_line(text):
    if not isinstance(text, str):
        return True
    lowered = text.lower()

    # şarkı/müzik dinleme ya da tv izleme odaklı cümleler
    if lowered.startswith("listening to:") or lowered.startswith("watching:"):
        return True

    # blog layout/template/meta yorumları
    if any(kw in lowered for kw in ["tagboard", "template", "blog layout", "click here", "bgcolor"]):
        return True

    return False

df = df[~df["text"].apply(is_useless_line)].copy()
df.reset_index(drop=True, inplace=True)
```

```python
import re

def contains_html_or_code(text):
    if not isinstance(text, str):
        return False

    # HTML etiketleri
    html_tag_pattern = r"<[^>]+>"

    # CSS stilleri
    css_style_pattern = r"style\s*=\s*['\"].*?['\"]"

    # JavaScript snippet'leri
    js_snippet_pattern = r"<script.*?>.*?</script>"

    return bool(re.search(html_tag_pattern, text)) or \
           bool(re.search(css_style_pattern, text)) or \
           bool(re.search(js_snippet_pattern, text, flags=re.DOTALL))

# uygula ve temizle
df = df[~df["text"].apply(contains_html_or_code)].copy()
df.reset_index(drop=True, inplace=True)
```

```python
import re

def is_useless_blog_line(text):
    if not isinstance(text, str):
        return False  # boş veriyle uğraşma
    lowered = text.strip().lower()

    # 1. sadece link veya urllink placeholder
    if re.fullmatch(r"(https?:\/\/\S+|www\.\S+)", lowered) or "urllink" in lowered:
        return True

    # 2. medya başlıkları (dinliyorum, izliyorum, oynuyorum gibi)
    if any(lowered.startswith(prefix) for prefix in ["listening to:", "watching:", "playing:", "currently listening to:", "now playing:"]):
        return True

    # 3. blog yapısal ifadeleri
    blog_meta_keywords = ["tagboard", "template", "layout", "click here", "bgcolor", "music player", "blogskin"]
    if any(kw in lowered for kw in blog_meta_keywords):
        return True

    # 4. sadece tarih, başlık ya da çizgi
    if re.fullmatch(r"\w+,\s\d{1,2}(st|nd|rd|th)?\s\w+", lowered):  # örnek: "Monday, 5th April"
        return True
    if lowered.strip() in ["---", "entry", "update"]:
        return True

    return False

# uygula ve temizle
df = df[~df["text"].apply(is_useless_blog_line)].copy()
df.reset_index(drop=True, inplace=True)
```

```python
combined_path = "/content/drive/MyDrive/datasets/combined_gender_text_dataset.pkl"

with open(combined_path, "rb") as f:
    df_combined = pickle.load(f)

df_merged = pd.concat([df, df_combined], ignore_index=True)

df_merged = df_merged[df_merged["text"].str.len() >= 15].copy()

min_count = df_merged["gender"].value_counts().min()
df_balanced = df_merged.groupby("gender").sample(n=min_count, random_state=63).reset_index(drop=True)

df_balanced
```
