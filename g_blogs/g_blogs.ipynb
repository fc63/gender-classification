```python
import zipfile

zip_path = "/content/drive/MyDrive/blogsgender/blogs.zip"
extract_path = "/content/blogs"

with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extractall(extract_path)
```

```python
import os
import pandas as pd
import xml.etree.ElementTree as ET

from bs4 import BeautifulSoup

data = []
folder_path = "/content/blogs/blogs"

for filename in os.listdir(folder_path):
    if not filename.endswith(".xml"):
        continue

    parts = filename.rsplit(".", 5)
    if len(parts) != 6:
        continue

    user_id, gender, age, industry, sign, _ = parts
    file_path = os.path.join(folder_path, filename)

    try:
        with open(file_path, 'r', encoding='utf-8', errors='ignore') as f:
            soup = BeautifulSoup(f.read(), 'html.parser')
            posts = soup.find_all("post")

            for post in posts:
                raw_text = post.get_text(strip=True)
                if raw_text:
                    data.append({
                        "text": raw_text,
                        "gender": gender
                    })

    except Exception as e:
        print(f"Yine hata ({filename}):", e)

df = pd.DataFrame(data)
print("Toplam satır:", len(df))
df.head()
```
