```python
import pandas as pd
import pickle
from sklearn.preprocessing import MinMaxScaler
import matplotlib.pyplot as plt

input_path = "/content/drive/MyDrive/datasets/nptel_gender_balanced.pkl"

with open(input_path, "rb") as f:
    df = pickle.load(f)
```

```python
input_path2 = "/content/drive/MyDrive/datasets/lehce1-engdataset.pkl"

with open(input_path2, "rb") as f2:
    df_lehce = pickle.load(f2)

df_lehce
```

```python
input_path3 = "/content/drive/MyDrive/datasets/rus_gender_en.pkl"

with open(input_path3, "rb") as f3:
    df_rus = pickle.load(f3)

df_rus
```

```python
import pandas as pd
import pickle

# sütun isimlerini eşitle
df_renamed = df.rename(columns={'transcription_normalised': 'text'})
df_lehce_renamed = df_lehce.rename(columns={'text': 'text'})
df_rus_renamed = df_rus.rename(columns={'text': 'text'})

# sadece gender ve text sütunlarını al ve birleştir
df_combined = pd.concat([
    df_renamed[['gender', 'text']],
    df_lehce_renamed[['gender', 'text']],
    df_rus_renamed[['gender', 'text']]
], ignore_index=True)

# pickle olarak drive'a kaydet
output_path = "/content/drive/MyDrive/datasets/combined_gender_text_dataset.pkl"
with open(output_path, "wb") as f:
    pickle.dump(df_combined, f)

print("Veri seti başarıyla birleştirildi ve kaydedildi:", output_path)
```
